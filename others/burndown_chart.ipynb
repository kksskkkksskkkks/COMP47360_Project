```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Data Preparation ---
total_days = 14
initial_points = 77+24

# 🎯 Planned Burndown - 直线
planned_remaining = np.linspace(initial_points, 0, total_days + 1)
days_planned = np.arange(len(planned_remaining))

# 🎯 每天实际工作小时数 (从 Day 1 开始填，没工作填 0)
daily_hours_worked = [
    0, 7, 10, 8, 3, 11, 10,   # Week 1 (周末填0)
    11, 10, 7, 0, 0, 0, 0    # Week 2
]

# 自动换算：累计消耗，得到每天剩余
actual_remaining = [initial_points]
for h in daily_hours_worked:
    actual_remaining.append(actual_remaining[-1] - h)

days_actual = np.arange(len(actual_remaining))

# --- 2. Plotting ---
plt.figure(figsize=(12, 7))

plt.plot(days_planned, planned_remaining,
         label='Planned Burndown',
         color='#1f77b4',
         linestyle='--',
         linewidth=2,
         marker='s',
         markersize=4,
         alpha=0.8)

plt.plot(days_actual, actual_remaining,
         label='Actual Burndown',
         color='#d62728',
         linestyle='-',
         linewidth=2.5,
         marker='o',
         markersize=6)

# --- 3. Chart Styling ---
plt.title('Sprint 1 Burndown Chart (2 Weeks)', fontsize=16, pad=20)
plt.xlabel('Time (Days)', fontsize=12)
plt.ylabel('Remaining Effort (Hours)', fontsize=12)

plt.xticks(np.arange(0, total_days + 1, 1))
plt.xlim(0, total_days)
plt.ylim(0, initial_points + 5)

plt.grid(True, linestyle=':', alpha=0.5)
plt.legend(loc='upper right', frameon=True, fontsize=11)

for week_start in [5, 12]:
    plt.axvspan(week_start, week_start + 2, color='gray', alpha=0.1)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Data Preparation ---
total_days = 14
initial_points = 77 + 24
today_day = 10  # ← 当前截止天数

# 🎯 Planned Burndown - 直线
planned_remaining = np.linspace(initial_points, 0, total_days + 1)
days_planned = np.arange(len(planned_remaining))

# 🎯 每天实际工作小时数 (从 Day 1 开始填，没工作填 0)
daily_hours_worked = [
    0, 7, 10, 8, 3, 11, 10,   # Week 1
    11, 10, 7, 0, 0, 0, 0     # Week 2
]

# 自动换算：累计消耗，得到每天剩余
actual_remaining = [initial_points]
for h in daily_hours_worked:
    actual_remaining.append(actual_remaining[-1] - h)

days_actual = np.arange(len(actual_remaining))

# --- 2. Plotting ---
fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(days_planned, planned_remaining,
        label='Planned Burndown',
        color='#1f77b4',
        linestyle='--',
        linewidth=2,
        marker='s',
        markersize=4,
        alpha=0.8)

ax.plot(days_actual, actual_remaining,
        label='Actual Burndown',
        color='#d62728',
        linestyle='-',
        linewidth=2.5,
        marker='o',
        markersize=6)

# --- 3. 未来区域 Hatch ---
ax.fill_between(
    [today_day, total_days],          # x 范围：today 到结束
    0, initial_points + 5,            # y 范围：底部到顶部
    hatch='///',                     # 45度斜线
    facecolor='none',                 # 无填充色
    edgecolor='lightgrey',            # 斜线颜色：浅灰
    linewidth=0.5,
    zorder=0                          # 置于最底层
)

# --- 4. Chart Styling ---
ax.set_title('Sprint 1 Burndown Chart (2 Weeks)', fontsize=16, pad=20)
ax.set_xlabel('Time (Days)', fontsize=12)
ax.set_ylabel('Remaining Effort (Hours)', fontsize=12)

ax.set_xticks(np.arange(0, total_days + 1, 1))
ax.set_xlim(0, total_days)
ax.set_ylim(0, initial_points + 5)

ax.grid(True, linestyle=':', alpha=0.5)

for week_start in [5, 12]:
    ax.axvspan(week_start, week_start + 2, color='gray', alpha=0.1)

# 今日截止竖线
ax.axvline(x=today_day, color='darkgrey', linestyle='--', linewidth=1.2, label=f'Plot Day (Day {today_day})')

ax.legend(loc='upper right', frameon=True, fontsize=11)
plt.tight_layout()
plt.show()
```
